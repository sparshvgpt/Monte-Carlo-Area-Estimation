# Estimating volumes using Monte Carlo Simulation

### Estimating volume of a unit sphere

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
N = 100000

x = np.random.uniform(-1,1,N)
y = np.random.uniform(-1,1,N)
z = np.random.uniform(-1,1,N)

inside = x**2 + y**2 + z**2 <= 1
volume_est = (np.sum(inside)/N) * 8

print(f"Estimated volume of sphere = {volume_est:.6f}, actual = 4/3*pi ≈ {4/3*np.pi:.6f}")
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x[inside], y[inside], z[inside], color='green', s=1)
plt.show()
```
